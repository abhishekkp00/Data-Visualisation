```python
import seaborn as sns    #sns is used for making good type of plots, advanced version of matplot
import matplotlib.pyplot as plt
import pandas as pd
```

### We use distribution plots to visualize the distribution of quantitative data.

Some plots that allow us to visualize the distribution of a data set. These plots are:

    distplot/histplot
    jointplot
    pairplot
    rugplot


```python
df = pd.read_csv('data/tips.csv')
df
```

```python
df['size'].unique()
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 20,kde = True) #kernel density estimator: skweness of data(left/right), see the curve on the plot 
plt.subplot(1,2,2)    #bins = 20 default
sns.histplot(df['tip'],bins = 20,kde = True)
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'scatter' )  #for comparing two data

#press shift + tab to see what all can we enter in the bracket for a field
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'reg')
```

histogram : to check the distribution of the data

```python
sns.pairplot(df,hue='size',palette='rainbow')  
```

```python
sns.rugplot(df['tip'])
```
