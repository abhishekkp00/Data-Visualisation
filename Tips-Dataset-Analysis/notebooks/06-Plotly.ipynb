### Interactive charts with Plotly

Plotly creates interactive charts with hover details, zooming and filtering. This notebook uses Plotly Express for a concise plotting workflow.

```python
import pandas as pd
import plotly.express as px
```

```python
tips = pd.read_csv('data/tips.csv')
tips.head()
```

```python
fig = px.scatter(
    tips,
    x="total_bill",
    y="tip",
    color="day",
    hover_data=["sex", "smoker", "size"],
    title="Tips compared with total bill",
)
fig.show()
```

```python
daily_tips = tips.groupby("day", as_index=False)["tip"].mean()
fig = px.bar(
    daily_tips,
    x="day",
    y="tip",
    title="Average tip by day",
    labels={"day": "Day", "tip": "Average tip"},
)
fig.show()
```
